# Exploratory Data Analysis - Fake News Dataset

This notebook explores the fake news dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
train_df = pd.read_csv('../data/train.csv')
train_df.head()
```

### Target Distribution
Let's see the balance between Fake (1) and Real (0) news.

```python
sns.countplot(x='label', data=train_df)
plt.title('Distribution of Fake vs Real News')
plt.show()
```

### Text Length Analysis

```python
train_df['text_length'] = train_df['text'].apply(lambda x: len(str(x).split()))
sns.histplot(data=train_df, x='text_length', hue='label', bins=50, kde=True)
plt.title('Word Count Distribution by Class')
plt.xlim(0, 2000)
plt.show()
```
